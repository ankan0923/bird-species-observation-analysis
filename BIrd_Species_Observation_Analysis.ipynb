Missing Values Handeling

```python
# fill null value with Unknown  
def fill_values(df,columns):
    for col in columns:
        df[col]=df[col].fillna('Unknown')
    return df

df=fill_values(df,['Sex','Distance','Site_Name'])    
```

```python
# sub_unit_code approximately 96% others two are only 2 value are missing so drop it
def drop_col(df, columns):
    df = df.drop(columns=columns)   # no inplace, returns new DataFrame
    return df

df = drop_col(df, ['Sub_Unit_Code', 'ID_Method', 'NPSTaxonCode'])
```

```python
# fill with average(mode) value
df['AcceptedTSN']=df['AcceptedTSN'].fillna(df['AcceptedTSN'].mode()[0])
```

Datatype Conversion

```python
df['Date']=pd.to_datetime(df['Date'])
df["Year"] = df["Date"].dt.year
```

Feature engineering

```python
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day
```

```python
#Season
def season(month):
    if month in [3,4,5]:
        return "Spring"
    elif month in [6,7,8]:
        return "Summer"
    elif month in [9,10,11]:
        return "Autumn"
    else:
        return "Winter"

df["Season"] = df["Month"].apply(season)
```

```python
from datetime import datetime, date
def add_duration_and_category(df, start_col="Start_Time", end_col="End_Time"):
    df[start_col] = pd.to_datetime(
        df[start_col], format="%H:%M:%S", errors="coerce").dt.time
    df[end_col] = pd.to_datetime(
        df[end_col], format="%H:%M:%S", errors="coerce").dt.time

    def calc_duration(start, end):
        if pd.isna(start) or pd.isna(end):
            return np.nan

        start_dt = datetime.combine(date.today(), start)
        end_dt = datetime.combine(date.today(), end)

        return (end_dt - start_dt).total_seconds() / 60

    df["Observation_Duration_Min"] = [
        calc_duration(start, end)
        for start, end in zip(df[start_col], df[end_col])]

    def categorize_time(t):
        if pd.isna(t):
            return "Unknown"

        hour = t.hour

        if 5 <= hour < 8:
            return "Early Morning"
        elif 8 <= hour < 12:
            return "Morning"
        elif 12 <= hour < 17:
            return "Afternoon"
        elif 17 <= hour < 21:
            return "Evening"
        else:
            return "Night"
    df["Time_of_Day"] = df[start_col].apply(categorize_time)
    return df

df = add_duration_and_category(df)
```

```python
def shannon_index(group):
    counts = group["Scientific_Name"].value_counts()
    proportions = counts / counts.sum()
    return -(proportions * np.log(proportions)).sum()

shannon = (
    df.groupby("Habitat")
      .apply(shannon_index)
      .reset_index(name="Shannon_Index"))

shannon
```

```python
#Habitat preference analysis
habitat_pref = pd.crosstab(
    df["Common_Name"],
    df["Habitat"])

habitat_pref["Total"] = habitat_pref.sum(axis=1)

habitat_pref["Forest_Pct"] = (
    habitat_pref.get("Forest", 0) /
    habitat_pref["Total"] * 100)

habitat_pref["Grassland_Pct"] = (
    habitat_pref.get("Grassland", 0) /
    habitat_pref["Total"] * 100)

habitat_pref.head()
```

```python
def preference(row):
    if row["Forest_Pct"] >= 70:
        return "Forest Preferred"
    elif row["Grassland_Pct"] >= 70:
        return "Grassland Preferred"
    else:
        return "Mixed Habitat"

habitat_pref["Habitat_Preference"] = habitat_pref.apply(preference,axis=1)
```
